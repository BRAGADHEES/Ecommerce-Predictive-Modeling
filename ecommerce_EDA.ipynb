```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
import numpy as np

# Load data
df = pd.read_csv('ecommerce_sales_data.csv')

# 1. Statistical Summary
print("--- Statistical Summary ---")
print(df.describe())

# 2. Correlation Matrix
# This shows how strongly variables like Sales and Profit relate to each other
plt.figure(figsize=(8, 6))
correlation = df[['Quantity', 'Sales', 'Profit']].corr()
sns.heatmap(correlation, annot=True, cmap='coolwarm', fmt=".2f")
plt.title('Correlation Heatmap: Sales vs Profit vs Quantity')
plt.show()

# 3. Distribution of Sales (Histogram)
plt.figure(figsize=(10, 5))
sns.histplot(df['Sales'], bins=30, kde=True, color='blue')
plt.title('Distribution of Sales Values')
plt.show()

# 4. Boxplot: Profitability by Category and Region (Multi-variate Analysis)
plt.figure(figsize=(12, 6))
sns.boxplot(data=df, x='Category', y='Profit', hue='Region')
plt.title('Profit Distribution by Category across Different Regions')
plt.legend(bbox_to_anchor=(1.05, 1), loc='upper left')
plt.show()

# 5. Scatter Plot: Sales vs Profit
plt.figure(figsize=(10, 6))
sns.scatterplot(data=df, x='Sales', y='Profit', hue='Category', alpha=0.6)
plt.title('Sales vs Profit Analysis by Category')
plt.show()
```
